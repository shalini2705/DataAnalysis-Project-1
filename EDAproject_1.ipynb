```python
# Replace ? with NaN
df = df.replace(" ?", None)
print(df.isnull().sum())
# Fill missing values with constant
df = df.fillna({'Workclass': 'Private','Occupation': 'Other-service','Native Country': 'United-States'})
print(df)
```

```python
# Cleaning
# EDA-4: Identify NaN values
df = df.replace(" ?", None )
print(df.isnull().sum())
print("_________________________")
# Approach-2.2: Fill with mean
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df)
```

```python
# Cleaning

# EDA-4: Identify NaN values
df = df.replace(" ?", None )
print(df.isnull().sum())
print("_________________________")
# Fill with median
df["Age"] = df["Age"].fillna(df["Age"].median())
print(df)
```

```python
# Cleaning

# EDA-4: Identify NaN values
df = df.replace(" ?", None )
print(df.isnull().sum())
print("_________________________")
# Fill with mode
df["Occupation"] = df["Occupation"].fillna(df["Occupation"].mode())
print(df)
```

```python
#import pandas as pd
#df = pd.read_csv("adult.csv")

# Cleaning
# Approach-1: Delete the records (rows)
df = df.dropna()
print(df)
print("_________________________")

# Approach-2: Delete the columns
df = df.dropna(axis=1)
print(df)
print("_________________________")
```

```python
#forward filling
df['Age']=df['Age'].ffill()
print(df)
df['Workclass'] = df['Workclass'].ffill()
print(df)
df['Ocuupation'] = df['Occupation'].ffill()
print(df)
df['Native Country'] = df['Native Country'].ffill()
print(df)
```

```python
#backward filling
df['Age']=df['Age'].bfill()
print(df)
df['Workclass'] = df['Workclass'].bfill()
print(df)
df['Ocuupation'] = df['Occupation'].bfill()
print(df)
df['Native Country'] = df['Native Country'].bfill()
print(df)
```

```python
df['Age']=df['Age'].interpolate()
print(df)
df['Workclass'] = df['Workclass'].interpolate()
print(df)
df['Ocuupation'] = df['Occupation'].interpolate()
print(df)

df['Native Country'] = df['Native Country'].interpolate()
print(df)
```

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3-Q1
print('Q1',Q1)
print('Q3',Q3)
print('IQR',IQR)
lower_bound = (Q1-1.5*IQR)
upper_bound = (Q3+1.5*IQR)
print(lower_bound)
print(upper_bound)
```

```python
iqr_outliers = df[(df['Age'] < lower_bound) | (df['Age'] > upper_bound)]
print("IQR Outliers:",len(iqr_outliers))
```

```python
#Z-Score Method
from scipy.stats import zscore
from scipy.stats.mstats import winsorize
df['Age_zscore'] = zscore(df['Age'])
z_outliers = df[df['Age_zscore'].abs() > 3]
print("Z-Score Outliers:", len(z_outliers))
```

```python
#Z-Score Trimming
df_trimmed = df[df['Age_zscore'].abs() <=3]
print(df_trimmed)
```

```python
#Winsorization
df['Age_winsorized'] = winsorize(df['Age'], limits=[0,0.5])
print(df.head())
```

```python
#using hours per week
Q1 = df['Hours per Week'].quantile(0.25)
Q3 = df['Hours per Week'].quantile(0.75)
IQR = Q3-Q1
print('Q1',Q1)
print('Q3',Q3)
print('IQR',IQR)
lower_bound = (Q1-1.5*IQR)
upper_bound = (Q3+1.5*IQR)
print(lower_bound)
print(upper_bound)
```

```python
#Identify IQR Outliers
iqr_outliers = df[(df['Hours per Week'] < lower_bound) | (df['Hours per Week'] > upper_bound)]
print("IQR Outliers:",len(iqr_outliers))
```

```python
#Z-Score
from scipy.stats import zscore
from scipy.stats.mstats import winsorize
df['Hours per Week_zscore'] = zscore(df['Hours per Week'])
z_outliers = df[df['Hours per Week_zscore'].abs() > 3]
print("Z-Score Outliers:", len(z_outliers))
```

```python
df_trimmed = df[df['Hours per Week_zscore'].abs() <=3]
print(df_trimmed)
```

```python
df['Hours per Week_winsorized'] = winsorize(df['Hours per Week'], limits=[0,0.5])
print(df.head())
```
